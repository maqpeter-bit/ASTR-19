```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 2*np.pi, 100)
#^ Defining an array here to be used as X
#print(x)
```

```python
def journalFunction(a, b):
    return a * np.exp(-b*x)
```

```python
a = 3
b = 4

y = journalFunction(a, b)

```

```python
plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y")

plt.savefig("CodingJournal2Session1Plot.pdf")
```

```python
plt.errorbar(y, np.random.normal(x))
```
